```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")

print("Path to dataset files:", path)
```

```python
import os

dataset_path = '/kaggle/input/brazilian-ecommerce'
dataset_contents = os.listdir(dataset_path)

print(f"Contents of {dataset_path}:")
for item in dataset_contents:
    print(item)
```

```python
import pandas as pd
customers = pd.read_csv(dataset_path + '/olist_customers_dataset.csv')
sellers = pd.read_csv(dataset_path + '/olist_sellers_dataset.csv')
order_reviews = pd.read_csv(dataset_path + '/olist_order_reviews_dataset.csv')
order_items = pd.read_csv(dataset_path + '/olist_order_items_dataset.csv')
products = pd.read_csv(dataset_path + '/olist_products_dataset.csv')
geolocation = pd.read_csv(dataset_path + '/olist_geolocation_dataset.csv')
product_category = pd.read_csv(dataset_path + '/product_category_name_translation.csv')
orders = pd.read_csv(dataset_path + '/olist_orders_dataset.csv')
order_payments = pd.read_csv(dataset_path + '/olist_order_payments_dataset.csv')
```

```python
all_tables = {
    'customers': customers,
    'sellers': sellers,
    'order_reviews': order_reviews,
    'order_items': order_items,
    'products': products,
    'geolocation': geolocation,
    'product_category': product_category,
    'orders': orders,
    'order_payments': order_payments
}
```

```python
for name, table in all_tables.items():
  print(f"Table: {name}, Shape: {table.shape}")
  print()
  display(table.info())
  print()
  display(table.describe())
  print()
  display(table.head(5))
  print()
  print('-'*25)
```

```python
order_reviews['review_creation_date'] = pd.to_datetime(order_reviews['review_creation_date'])
order_reviews['review_answer_timestamp'] = pd.to_datetime(order_reviews['review_answer_timestamp'])
order_items['shipping_limit_date'] = pd.to_datetime(order_items['shipping_limit_date'])

date_order = ['order_purchase_timestamp', 'order_approved_at','order_delivered_carrier_date',
              'order_delivered_customer_date', 'order_estimated_delivery_date']
for d in date_order:
  orders [d] = pd.to_datetime(orders[d])
```

```python
for name, table in all_tables.items():
    print(f"--- Analysis for Table: {name} ---")

    # Check for Null Values
    null_counts = table.isnull().sum()
    if null_counts.sum() > 0:
        print("Null values per column:")
        print(null_counts[null_counts > 0])
    else:
        print("No null values found.")

    # Check for Duplicate Data
    duplicate_count = table.duplicated().sum()
    print(f"Duplicate rows: {duplicate_count}")

    print('\n' + '-'*30 + '\n')
```

- Review title and messages - null values are acceptable
- Products table has some null values which need further investigation
- Duplicate rows in geolocation are acceptable
- Orders table needs more investigation

```python
print("--- Products with Null Category Names ---")
display(products[products['product_category_name'].isnull()])

print("/n --- Products with Null dimensions ---")
display(products[products['product_weight_g'].isnull()])

print("\n--- Orders with Null Delivery Dates ---")
# Checking if null delivery dates correlate with specific order statuses
display(orders[orders['order_delivered_customer_date'].isnull()]['order_status'].value_counts())

print("\n--- Sample of Orders with Null Delivery Dates ---")
display(orders[orders['order_delivered_customer_date'].isnull()])
```

- Orders Table: The missing order_delivered_customer_date values are almost entirely linked to orders that are shipped (1,107), canceled (619), or unavailable (609). There are only 8 orders marked as 'delivered' that lack a timestamp, which may be data entry errors.
- Products Table: Rows missing a product_category_name also lack other metadata like product_name_length and product_description_length. However, they still contain physical attributes like weight and dimensions with only 1 exception.
